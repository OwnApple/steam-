```python
import pandas as pd
import numpy as np

def preprocess_steam_data(file_path):
    print("🚀 开始数据预处理...")
    
    # 1. 加载数据
    # 注意：根据你的实际文件格式，可能需要调整 encoding 或 separator (如 sep='\t')
    df = pd.read_excel(file_path)
    df.columns = df.columns.str.strip()
    print(f"原始数据维度: {df.shape}")

    # 2. 宽表转长表 (Wide to Long)
    # 我们把用户的基础特征（比如用户名、页面评价）和对应的游戏数据剥离出来，然后纵向拼接
    base_cols = ['用户名', '评论', '评论页面', '有价值数', '欢乐数']
    
    # 假设你的数据里最多有 3 个游戏（根据你的 head 数据推测，如果有更多，修改 range 的范围即可）
    long_dfs = []
    for i in range(1, 4):
        game_col = f'游戏{i}'
        time_col = f'时长{i}'
        achieve_col = f'成就{i}'
        
        # 如果列存在，则提取并重命名为统一的列名
        if game_col in df.columns:
            temp_df = df[base_cols + [game_col, time_col, achieve_col]].copy()
            temp_df.rename(columns={
                game_col: '游戏名称',
                time_col: '游玩时长',
                achieve_col: '成就数据'
            }, inplace=True)
            long_dfs.append(temp_df)
            
    # 将拆分后的数据纵向拼接成真正的交互长表
    long_df = pd.concat(long_dfs, ignore_index=True)
    
    # 3. 清洗无效数据
    # 如果游戏名称是 NaN（即用户没有 游戏2 或 游戏3），直接删除这些空行
    long_df.dropna(subset=['游戏名称'], inplace=True)
    print(f"转化为长表并清洗空值后的维度: {long_df.shape}")

    # 4. 生成唯一 ID (数字化)
    # 算法只能处理数字，不能处理中英文名字，所以我们要给用户和游戏发“身份证”
    # category().cat.codes 会自动为每个唯一值分配一个从 0 开始的整数，我们 +1 让 ID 从 1 开始
    long_df['user_id'] = long_df['用户名'].astype('category').cat.codes + 1
    long_df['game_id'] = long_df['游戏名称'].astype('category').cat.codes + 1

    # 5. 构建评分 (Rating) 规则
    # 这是协同过滤 (ItemCF) 的灵魂。我们把文本评价和时长结合，量化为 1-5 分
    def calculate_rating(row):
        eval_status = str(row['评论页面'])
        playtime = float(row['游玩时长']) if pd.notna(row['游玩时长']) else 0.0
        
        if eval_status == '推荐':
            return 5
        elif eval_status == '不推荐':
            # 即使不推荐，但如果玩了超过 10 小时，说明游戏本身有一定的吸引力（比如是黑粉），给个基础分 3 分
            # 玩得少还不推荐，直接 1 分
            return 3 if playtime > 10 else 1 
        else:
            # 默认中立分
            return 3
            
    long_df['rating'] = long_df.apply(calculate_rating, axis=1)

    # 6. 整理最终需要导出的表结构，对标课件里的 users 和 ratings 格式
    
    # 构建 Users 表 (这里可以顺便提取特征给未来的 DSSM 双塔模型用)
    users_df = long_df[['user_id', '用户名']].drop_duplicates()
    
    # 构建 Games 表
    games_df = long_df[['game_id', '游戏名称']].drop_duplicates()
    
    # 构建 Ratings 交互表
    ratings_df = long_df[['user_id', 'game_id', 'rating', '游玩时长']]

    print("✅ 数据预处理完成！")
    return users_df, games_df, ratings_df, long_df

# ================= 测试运行 =================
if __name__ == "__main__":
    # 假设你的文件叫 steam_reviews.csv
    users, games, ratings, full_data = preprocess_steam_data('明末评论.xls')
    
    # 导出为 CSV，方便下一步利用 PyMySQL 导入 MySQL 数据库
    users.to_csv('./data/users_cleaned.csv', index=False)
    games.to_csv('./data/games_cleaned.csv', index=False)
    ratings.to_csv('./data/ratings_cleaned.csv', index=False)
    full_data.to_csv('./data/full_features_for_dssm.csv', index=False)
    print("可以在这里取消注释，执行数据读取和导出")
```
